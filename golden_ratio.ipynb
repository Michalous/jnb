# Golden Ratio

## Introduction

The golden ratio (φ ≈ 1.618) is a mathematical constant that appears frequently in nature, art, and architecture. This notebook simulates how seeds grow on a sunflower, demonstrating how the golden ratio creates the characteristic spiral patterns seen in nature.

The visualization shows seeds being "planted" sequentially along these spirals, starting from the center and working outward. By comparing the golden angle spiral with a regular angle like π, we can see why nature favors the golden ratio for optimal seed distribution.

## Implementation

1. **Golden Ratio Calculation**: φ is computed using the quadratic formula solution to φ² - φ - 1 = 0.

2. **Spiral Generation**: The `draw_dots` function places 300 points along a logarithmic spiral using polar coordinates. Each point's angle is incremented by the specified angle parameter, while the radius increases linearly.

Key parameters:
- **Number of points**: 300 for clear pattern visualization
- **Radius scaling**: i/500 creates a gradual outward expansion
- **Angle increment**: Default is φ (golden angle), compared with π for contrast

The visualization uses matplotlib scatter plots to render the spiral patterns, with point sizes adjusted for visual clarity.

```python
import math
import matplotlib.pyplot as plt
```

```python
phi = (1 + 5**0.5) / 2 # Golden Ratio
def draw_dots(angle = phi):
    inc = angle
    x, y = [], []
    for i in range(1, 300):
        x.append(math.cos((2 * math.pi) * angle) * i/500)
        y.append(math.sin((2 * math.pi) * angle) * i/500)
        angle += inc
    return [x, y]
```

```python
x, y = draw_dots()
plt.figure(figsize=(10, 10))
plt.scatter(x, y, s=20);
```

```python
x, y = draw_dots(math.pi)
plt.figure(figsize=(10, 10))
plt.scatter(x, y, s=10);
```

## Results

The visualizations above simulate the growth of sunflower seeds using different angular increments.

**First Plot (Golden Angle)**: Using φ ≈ 1.618 as the angle increment creates the characteristic sunflower seed pattern. Each "seed" is placed at the optimal golden angle from the previous one, resulting in two interlocking spirals (clockwise and counterclockwise) that efficiently fill the space. This is exactly how real sunflowers grow their seeds.

**Second Plot (π Angle)**: Using π ≈ 3.1416 creates a more regular but less efficient distribution. The seeds cluster more tightly and don't achieve the same space-filling efficiency as the golden spiral.
